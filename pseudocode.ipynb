# Algorithm and Pseudocode Comparison

This section presents the Value Iteration algorithm implemented in this project and explains how it relates to the classical version described in *Artificial Intelligence: Foundations and Computational Agents* (Poole and Mackworth, Figure 9.16).

## 1. Original Textbook Pseudocode (Figure 9.16)

The standard synchronous Value Iteration algorithm is defined as follows:

```text
procedure ValueIteration(S, A, P, R, γ)

    initialize V₀[s] arbitrarily for all s ∈ S
    k ← 0

    repeat
        k ← k + 1

        for each state s ∈ S do
            Vₖ[s] ← maxₐ [ R(s,a) + γ ∑ₛ' P(s'|s,a) Vₖ₋₁[s'] ]

    until convergence

    for each state s ∈ S do
        π[s] ← argmaxₐ [ R(s,a) + γ ∑ₛ' P(s'|s,a) Vₖ[s'] ]

    return Vₖ, π

## 2. Implementation Used in This Project

The implementation in this project follows the same Bellman update principle but is adapted for practical use and clarity in code.

procedure ValueIteration(S, A, P, R, γ, θ)

    initialize V[s] ← 0 for all s ∈ S

    repeat
        Δ ← 0

        for each state s ∈ S do

            if s has no available actions then
                continue

            v* ← -∞

            for each action a ∈ A(s) do

                q ← 0

                for each (p, s') ∈ P(s,a) do
                    q ← q + p [ R(s,a,s') + γ V[s'] ]

                v* ← max(v*, q)

            Δ ← max(Δ, |V[s] - v*|)
            V[s] ← v*

    until Δ < θ

    for each state s ∈ S do

        if s has no available actions then
            π[s] ← None
            continue

        π[s] ← argmaxₐ ∑ₛ' P(s'|s,a) [ R(s,a,s') + γ V[s'] ]

    return V, π

## 3. Differences

The implementation differs from the textbook version in several key ways:


In-place updates:
    The textbook uses separate value functions ( $V_k$ ) and ( $V_{k-1}$ ), while this implementation updates a single value function directly. This simplifies the implementation and can improve convergence speed.


Explicit convergence criterion:
    Instead of an abstract termination condition, this implementation uses a threshold ( $\theta$ ) and stops when the maximum change in value falls below this threshold.

Reward structure:
    The textbook uses ( $R(s,a)$ ), whereas this implementation uses ( $R(s,a,s')$ ), allowing rewards to depend on the resulting state.

Terminal states:
    States without available actions are explicitly handled by skipping updates and assigning no policy.

Policy extraction:
    The policy is computed after convergence, maintaining a clear separation between value computation and decision extraction.
